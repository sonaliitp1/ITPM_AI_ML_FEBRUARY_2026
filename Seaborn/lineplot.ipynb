```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
# Sample Dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Maths": [78, 85, 90, 76, 88, 92, 95, 70],
    "Science": [80, 89, 94, 72, 86, 91, 96, 68],
    "English": [75, 82, 88, 70, 84, 90, 93, 65],
    "Attendance": [85, 90, 95, 80, 88, 92, 97, 75]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.lineplot(x="Student", y="Maths", data=df)
plt.title("Line Plot")
plt.show()
```

```python
sns.barplot(x="Student", y="Science", data=df)
plt.title("Bar Plot")
plt.show()
```

```python
plt.hist(df["Attendance"], bins=5)
plt.title("Histogram - Attendance Distribution")
plt.xlabel("Attendance")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(df["Attendance"], bins=5)
plt.title("Histogram")
plt.show()
```

```python
sns.histplot(df["Maths"], bins=5, kde=True) # kernal density estimation
plt.title("Maths Marks Distribution with KDE")
plt.show()
```

```python
df[["Maths", "Science", "English"]].plot(kind="hist", bins=5, alpha=0.5)
plt.title("Combined Histogram")
plt.show()
```

```python
columns = ["Maths", "Science", "English", "Attendance"]

for col in columns:
    sns.histplot(df[col], bins=5, kde=True)
    plt.title(f"{col} Distribution")
    plt.show()
```

```python
sns.scatterplot(x="Maths", y="Science", data=df)
plt.title("Scatter Plot")
plt.show()
```

```python
sns.kdeplot(df["Maths"], fill=True)
plt.title("KDE Plot")
plt.show()
```

```python
sns.heatmap(df[["Maths", "Science", "English", "Attendance"]].corr(), annot=True)  # annot=True Show numbers inside the boxes
plt.title("Heatmap")
plt.show()
```
