```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y,marker="*",color="green",lw=4,mec="red",mfc="blue",markersize="10")
plt.title("Line Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

plt.bar(categories, values, color='purple')
plt.title("Bar Chart",fontdict={"color":"blue","fontsize":20,"fontfamily":"Arial"})
plt.xlabel("Categories",fontdict={"color":"red","fontsize":10,"fontfamily":"Arial"})
plt.ylabel("Values",fontdict={"color":"yellow","fontsize":10,"fontfamily":"Arial","fontweight":"bold"})
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
qty = np.array([15,45,60,70,80,33])
fruits = ["Apple","Banana","Orange","Pineapple","Kiwi","Mango"]
plt.pie(qty,labels=fruits)
plt.title("Fruits Quantity")
plt.legend()
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([11,22,33,44,55])
plt.scatter(x,y)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85]

sns.kdeplot(marks, fill=True) #It shows the concentration (density) of marks.
plt.title("KDE Plot")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95]

sns.histplot(marks, bins=6, kde=True)

plt.title("Histogram with KDE") # kDE probability distribution of a dataset.
                                # shows the shape of the data distribution.
plt.show()
```
