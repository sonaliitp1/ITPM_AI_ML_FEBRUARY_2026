```python
import seaborn as sns 
import matplotlib.pyplot as plt
import pandas as pd 

data = sns.load_dataset("titanic")
data
```

```python
sns.scatterplot(data=data, x="age", y="fare")
plt.show()
```

```python
sns.scatterplot(data=data,
                x="age",
                y="fare",
                hue="survived", style="class")
plt.show()
```

```python
sns.barplot(data=data,
            x="class",
            y="fare")
plt.show()
```

```python
sns.histplot(data=data,
             x="age",
             kde=True)
plt.show()
```

```python
sns.boxplot(data=data,
            x="class",
            y="fare")
plt.show()
```

```python
sns.boxplot(data=data,
            x="class",
            y="fare",
            hue="sex")
plt.show()
```

```python
data1 = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Maths": [78, 85, 90, 76, 88, 92, 95, 70],
    "Science": [80, 89, 94, 72, 86, 91, 96, 68],
    "English": [75, 82, 88, 70, 84, 90, 93, 65],
    "Attendance": [85, 90, 95, 80, 88, 92, 97, 75],
    "age":[20,22,23,24,25,26,27,28]
}

df = pd.DataFrame(data1)

print(df)
sns.heatmap(df[["Maths", "Science", "English", "Attendance","age"]].corr(), annot=True)  # annot=True Show numbers inside the boxes
plt.title("Heatmap")
plt.show()
```

```python
corr = data.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
columns = ["Maths", "Science", "English", "Attendance"]

for col in columns:
    sns.histplot(df[col], bins=5, kde=True)
    plt.title(f"{col} Distribution")
    plt.show()
```
