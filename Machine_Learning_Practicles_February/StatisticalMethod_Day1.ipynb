```python
import pandas as pd 

data ={
    "age":[23,22,24,25,26,27],
    "salary":[34000,44000,45678,67000,23000,29000],
    
}

df = pd.DataFrame(data)
df
```

```python
print(df.mean())
print(df.median())
print(df.mode())
```

```python
print(df.max())
print(df.min())
# variance
print(df.var())
# standard deviation
print(df.std())
# Skewness
print(df.skew())
# Kurtosis 
print(df.kurt())
Q1 = df.quantile(0.25)
Q2 = df.quantile(0.50)
Q3 = df.quantile(0.75)
print("Q1 :-",Q1)
print("Q2 :-",Q2)
print("Q3 :-",Q3)
```
