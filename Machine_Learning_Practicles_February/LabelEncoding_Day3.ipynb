```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.DataFrame({
    "city":["Pune","Mumbai","Pune","Delhi","Pune","Mumbai"],
    "name":["Abc","Pqr","Xyz","Mnq","Tre","Aaa"]
})

le_object = LabelEncoder() # create object of LabelEncoder

data["city_new"] = le_object.fit_transform(data['city'])

print(data)
```
