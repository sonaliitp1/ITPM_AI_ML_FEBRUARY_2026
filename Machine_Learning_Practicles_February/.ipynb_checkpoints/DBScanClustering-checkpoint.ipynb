```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler


# Customer data
X = np.array([
    [15, 39],
    [16, 45],
    [17, 42],
    [70, 80],
    [75, 85],
    [72, 88],
    [30, 10],
    [234,567]
])


# Feature scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# Create DBSCAN model
dbscan = DBSCAN(
    eps=0.8,      #eps defines how close two points should be to be considered neighbors(eps as a circle radius around 		  #each data point.
    min_samples=3  #Minimum number of points required inside the eps radius to create a cluster.
)
# Train model
clusters = dbscan.fit_predict(X_scaled)


print("Cluster Labels:")
print(clusters)


# Visualization
plt.scatter(
    X[:,0],
    X[:,1],
    c=clusters,
    s=100
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("DBSCAN Customer Segmentation")

plt.show()
```
