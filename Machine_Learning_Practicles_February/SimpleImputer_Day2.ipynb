```python
import numpy as np 

from sklearn.impute import SimpleImputer
```

```python
data = np.array([
    [3,4,np.nan],
    [4,np.nan,5],
    [6,7,np.nan]
])

print("Old Data:- ")

print(data)

# create object of class SimpleImputer

obj = SimpleImputer(strategy="mean")

new_data = obj.fit_transform(data)

print("New Data after transform:-")

print(new_data)











```
