```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
X = np.array([
    [78,90],   #Initially every point is a separate cluster:Algorithm calculates distances.
    [75,85],
    [17,42],
    [70,80],
    [16,45],
    [15,39]
])
```

```python
# Dendrogram
plt.figure(figsize=(8,5))

linkage_data = linkage(  
    X,
    method="ward"    # Merge clusters that create the smallest increase in variance.
)                    
                    #  Calculate Linkage Matrix
                     # Linkage decides how to measure distance between clusters.

```

```python
dendrogram(linkage_data) #Draw Dendrogram(creates the tree diagram)

plt.title("Customer Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Distance")

plt.show()
```

```python
#Hierarchical Model

hc = AgglomerativeClustering(
    n_clusters=2,
    metric="euclidean",
    linkage="ward"  #Ward's method tries to combine groups whose members are relatively similar.
)
```

```python
clusters = hc.fit_predict(X) #Assign Cluster Labels
print("Cluster Labels:")
print(clusters)
```

```python
plt.scatter(
    X[:,0],
    X[:,1],
    c=clusters,
    s=100
)

plt.xlabel("Income")
plt.ylabel("Spending Score")

plt.title(
    "Hierarchical Customer Segmentation"
)

plt.show()
```
