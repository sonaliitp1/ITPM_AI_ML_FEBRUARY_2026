```python
import pandas as pd 
from sklearn.preprocessing import MinMaxScaler
```

```python
data ={
    "squareFoot":[1200,1500,1800,1000,1600],
    "bedrooms":[3,2,1,3,1],
    "price":[120000,320000,340000,56000,76000]
}

df = pd.DataFrame(data)

# create the object of minmaxscaler

model = MinMaxScaler()

scaled_data = model.fit_transform(df)

print(scaled_data.round(1))
```
