```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
dataset = pd.read_csv("Mall_Customers.csv")
```

```python
dataset
```

```python
# Select Features
X = dataset.iloc[:, [3, 4]].values
# Annual Income, Spending Score
```

```python
X
```

```python
# Train KMeans++
# ----------------------------
kmeans = KMeans(
    n_clusters=5,
    init='k-means++', #init='k-means++' Instead of choosing centroids randomly,KMeans++ chooses smart initial centroids.
    random_state=42, #random_state=42 Without it every run gives different clusters.,you always get the same output.
    n_init=10 #10 different initializations and chooses the one having the lowest WCSS.
)
```

```python
clusters = kmeans.fit_predict(X) # fit() = Train the KMeans model.It finds 5 centroids
                                 #predict()=  Assign every customer to its nearest centroid.
```

```python
# Add cluster column to dataset
dataset["Cluster"] = clusters # Store Cluster Number
```

```python
# ----------------------------
# Display Customers in Each Cluster
# ----------------------------
for i in range(5):  #Display Cluster Data
    print("\n" + "="*60)
    print(f"Cluster {i}")
    print("="*60)

    cluster_data = dataset[dataset["Cluster"] == i] #This is called Boolean Indexing.

    print(cluster_data[[   #Prints only those columns.
        "CustomerID",
        "Gender",
        "Age",
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]])
```

```python
# ----------------------------
# Add New Customer
# ----------------------------

income = float(input("\nEnter Annual Income (k$): "))
score = float(input("Enter Spending Score (1-100): "))
```

```python
new_customer = [[income, score]] #Create New Customer (KMeans expects a 2D array.)

cluster = kmeans.predict(new_customer)[0] # Predict Cluster KMeans calculates distance from

```

```python
print("\nNew Customer belongs to Cluster:", cluster) #Show Assigned Cluster
```
