```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# --------------------------------------------------
# Step 1: Create Dataset
# --------------------------------------------------

data = {
    "Math": [90, 80, 70, 60, 50],
    "Physics": [85, 75, 65, 55, 45],
    "Chemistry": [88, 78, 68, 58, 48],
    "Biology": [80, 70, 60, 50, 40],
    "English": [75, 65, 55, 45, 35]
}
```

```python
df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
# --------------------------------------------------
# Step 2: Check Dataset Shape
# --------------------------------------------------

print("\nOriginal Shape:")
print(df.shape)
```

```python
# --------------------------------------------------
# Step 3: Standardize Data
# --------------------------------------------------

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)


print("\nStandardized Data:")
print(scaled_data)
```

```python
# --------------------------------------------------
# Step 4: Apply PCA
# Reduce 5 Features into 2 Features
# --------------------------------------------------

pca = PCA(n_components=2)


principal_components = pca.fit_transform(
    scaled_data
)
```

```python
# --------------------------------------------------
# Step 5: Create PCA DataFrame
# --------------------------------------------------

pca_df = pd.DataFrame(
    principal_components,
    columns=[
        "Principal_Component_1",
        "Principal_Component_2"
    ]
)
```

```python
print("\nAfter PCA Transformation:")
print(pca_df)
```

```python
print("\nNew Shape After PCA:")
print(pca_df.shape)
```

```python
# --------------------------------------------------
# Step 7: Explained Variance Ratio
# --------------------------------------------------

variance = pca.explained_variance_ratio_


print("\nExplained Variance Ratio:")
print(variance)


print("\nTotal Information Retained:")
print(sum(variance) * 100, "%")
```
