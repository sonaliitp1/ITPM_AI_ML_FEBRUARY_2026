```python
from sklearn.feature_extraction import DictVectorizer
```

```python
data =[
    {"age":30,"gender":"Male"},
    {"age":40,"gender":"Female"},
    {"age":50,"gender":"Male"}
]
```

```python
model = DictVectorizer()
```

```python
new_data = model.fit_transform(data)
```

```python
features = model.get_feature_names_out()

print(features)
print(new_data.toarray())
```
