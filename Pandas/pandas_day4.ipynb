```python
import pandas as pd

bankaccount ={
    "cust_id":[1122,2345,6677,8843,8899,1122],
    "cust_name":["John","Martin","Scott","Joe","Harry","John"],
    "balance":[56000,78000,99000,23456,48976,56000],
    "acct_type":["saving","current","saving","current","saving","saving"],
    "city":["Pune","Mumbai","Nashik","Pune","Nashik","Pune"]

} 

data = pd.DataFrame(bankaccount)
print(data)









```

```python
# only Aggregate Function
print("Max Balance :-",data["balance"].max())
print("Min Balance :-",data["balance"].min())
print("Sum of Balance :-",data["balance"].sum())
print("Count :-",data["balance"].count())
print("Average :-",data["balance"].mean())
```

```python
data.groupby("acct_type").count()
# data.groupby("acct_type")["balance"].count()
```

```python
# print average  of each account type-- saving current
data.groupby("acct_type")["balance"].mean()
```

```python
# Multiple Aggregation 

data.groupby("acct_type")["balance"].agg(['sum','max','min'])
```

```python
# print higehest balance of "Pune" city 

# result = data.groupby("city")["balance"].max()
# print(result['Pune'])

data.groupby("city")["balance"].max()[["Pune"]]
```

```python
# print higehst balance of each city with employee name

result = data.groupby("city").max()
print(result[['cust_name','balance']])
```
