```python
import pandas as pd

bankaccount ={
    "cust_id":[1122,2345,6677,8843,8899,1122],
    "cust_name":["John","Martin","Scott","Joe","Harry","John"],
    "balance":[56000,78000,99000,23456,48976,56000],
    "acct_type":["saving","current","saving","current","saving","saving"],
    "city":["Pune","Mumbai","Nashik","Pune","Nashik","Pune"]

} 

data = pd.DataFrame(bankaccount)
print(data)

data1 = pd.DataFrame({
    "name":["Raj","Soham","Riya","Pranav"],
    "age":[20,21,22,23],
    "city":["Pune","Mumbai","Nashik","Nagpur"]

},index=["A","B","C","D"])

data1
```

```python
# loc(string and integer based indexing) and iloc(integer based indexing)

#  select row 
print(data1.loc["B"])

# select multiple rows

print(data1.loc[["A","B"]])

# select specific data

print(data1.loc["A","city"])

# select rows as per given range

print(data1.loc["A":"D"])
```

```python
# select specific rows and specific column 

print(data1.loc["A":"C",["name","city"]])

# boolean filtering

print(data1.loc[data1["age"]>15])

# number based indexing in loc function

print(data.loc[0])

# multiple rows 

print(data.loc[1:4])
```

```python
# iloc function 

# select specific row 

print(data.iloc[3])

# select multiple rows 

print(data.iloc[[0,3,4]])

# select specific data

print(data.iloc[1,3])

# select range of rows and columns

print(data.iloc[0:2,0:2])

# select all rows but print columns as per range

print(data.iloc[:,0:2])

# print alternate rows

print(data.iloc[::2])

```
