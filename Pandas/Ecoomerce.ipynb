```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt


```

```python
# load file
data = pd.read_csv("order.csv")
data
```

```python
# show first 5 records
data.head()
# show data information
data.info()
# check null values count
data.isnull().sum()
```

```python
# Remove missing values
data = data.dropna()

# drop duplicates 
data = data.drop_duplicates()
```

```python
# Add new column
data["TotalAmount"] = np.multiply(data["Quantity"], data["Price"])
data
```

```python
# Filter Data 
# print("\n========== ORDERS FROM PUNE ==========")
pune_orders = data[data["City"] == "Pune"]
print(pune_orders)
```

```python
print("\n========== DELIVERED ORDERS ==========")
delivered_orders = data[data["Status"] == "Delivered"]
print(delivered_orders)
```

```python
print("\n========== SORT BY TOTAL AMOUNT ==========")
sorted_orders = data.sort_values(by="TotalAmount", ascending=False)
print(sorted_orders)
```

```python
print("\n========== PRODUCTS BY CATEGORY ==========")
category_orders = data.groupby("Category")["Product"].count()
print(category_orders)
```

```python
print("\n========== ORDERS BY PAYMENT METHOD ==========")
payment_orders = data.groupby("Payment")["OrderID"].count()
print(payment_orders)
```

```python
data.to_csv("processed_orders.csv", index=False)

print("\nProcessed dataset saved successfully.")
```

```python
# Visualization 
# --------------------------------------------
# 1. Orders by Category
# --------------------------------------------
plt.figure(figsize=(6,4))
sns.countplot(data=data, x="Category")
plt.title("Orders by Category")
plt.xticks(rotation=30)
plt.show()
```

```python
# 2. Payment Method
# --------------------------------------------
plt.figure(figsize=(6,4))
sns.countplot(data=data, x="Payment")
plt.title("Payment Method Used")
plt.show()
```

```python
#4. City-wise Orders
# --------------------------------------------
plt.figure(figsize=(7,4))
sns.countplot(data=data, x="City")
plt.title("Orders by City")
plt.xticks(rotation=45)
plt.show()
```

```python
#5. Quantity Purchased by Category
# --------------------------------------------
plt.figure(figsize=(7,4))
sns.barplot(data=data, x="Category", y="Quantity")
plt.title("Quantity Purchased by Category")
plt.xticks(rotation=30)
plt.show()
```

```python
#6. Total Amount by Payment Method
# --------------------------------------------
plt.figure(figsize=(6,4))
sns.barplot(data=data, x="Payment", y="TotalAmount")
plt.title("Total Amount by Payment Method")
plt.show()
```
