```python
import pandas as pd 

data = pd.DataFrame({
    "name":["Raj","Riya","Pranav","Simran","Siya"],
    "salary":[45678,56000,67000,76543,34000],
    "department":["IT","Hr","Sales","IT","Hr"],
    "age":[22,23,24,25,26],
    "grade":["A","B","A","C","C"]
})

print(data)


```

```python
grade_dict={
    "A":"Excellent",
    "B":"Good",
    "C":"Average"
}
def fun(grade):
    if grade=="A":
        return "Excellent"
    elif grade == "B":
        return "Good"
    else:
        return "Average"
# data["performance"] = data["grade"].map(grade_dict)

data["performance"] = data["grade"].map(fun)
print(data)
```

```python
data["age"] = data["age"].map(lambda x : x+5 )
data["newname"] = data["name"].map(str.upper)
data["name"]  = data["name"].map(lambda x:str.upper(x))
print(data)
```

```python
# apply 

df = pd.DataFrame({
    "nums":[1,2,3,4,5,6]
})

print(df)

def sq(n):
    return n*n

df["square"] = df["nums"].apply(sq)
df["cube"] = df["nums"].apply(lambda x:x**3)
print(df)
```

```python
df1 = pd.DataFrame({
    "Math":[78,89,90],
    "Science":[56,47,33]
})

print(df1)

def totalmarks(row):
    return row["Math"]+row["Science"]

df1["total_marks"] = df1.apply(totalmarks,axis=1)

print(df1)

# status = pass / fail -- math and science >50 -- pass otherwise -- fail
```
